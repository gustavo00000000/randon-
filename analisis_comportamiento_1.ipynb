```python
import pandas as pd

# 1. Creamos datos falsos de "comportamiento de clientes"
datos = {
    'cliente_id': [1, 2, 3, 4, 5],
    'compras_mes': [10, 2, 15, 1, 20],
    'minutos_en_app': [50, 10, 120, 5, 200],
    'genero': ['F', 'M', 'F', 'M', 'F']
}

df = pd.DataFrame(datos)

# 2. Análisis de comportamiento: ¿Quiénes son los clientes más activos?
print("--- Clientes con más de 10 compras ---")
print(df[df['compras_mes'] > 10])

# 3. Calcular el promedio de tiempo en la app por género
print("\n--- Promedio de tiempo en App por Género ---")
print(df.groupby('genero')['minutos_en_app'].mean())
```
